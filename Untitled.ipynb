```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

df = pd.read_excel('expenses.xlsx', sheet_name=0)

# 让 pandas 显示所有行
pd.set_option('display.max_rows', None)

df['date'] = pd.to_datetime(df['date'])
df   # 这里 head() 也可以直接改成 df，但留着也行
```

```python
# 数据基本信息
df.info()

# 描述性统计（只看金额列）
df['amount'].describe()
```

```python
income_total = df[df['category'] == '收入']['amount'].sum()
expense_total = df[df['category'] != '收入']['amount'].sum()
net = income_total - expense_total

print(f"总收入：{income_total:.2f} 元")
print(f"总支出：{expense_total:.2f} 元")
print(f"净结余：{net:.2f} 元")
```

```python
# 先筛出非收入的支出数据
expense_df = df[df['category'] != '收入']

# 按类别汇总
category_expense = expense_df.groupby('category')['amount'].sum().sort_values(ascending=False)
print(category_expense)

# 画饼图
plt.figure(figsize=(8, 6))
plt.pie(category_expense, labels=category_expense.index, autopct='%1.1f%%')
plt.title('8月支出类别占比')
plt.show()
```

```python
daily_expense = expense_df.groupby('date')['amount'].sum()

plt.figure(figsize=(10, 4))
daily_expense.plot(kind='line', marker='o')
plt.title('每日支出趋势')
plt.xlabel('日期')
plt.ylabel('金额（元）')
plt.grid(True)
plt.show()
```
